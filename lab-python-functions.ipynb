````python
lst = [1,2,3,3,3,3,4,5]
def get_unique_list(lst):
    """
    Toma una lista como argumento y devuelve una nueva lista con elementos únicos de la primera lista.

    Parámetros:
    lst (lista): la lista de entrada.

    Devoluciones:
    lista: una nueva lista con elementos únicos de la lista de entrada.
    """
    return list(set(lst))
````

````python
string = "Hello World"
def count_case(string):
    """
    Returns the number of uppercase and lowercase letters in the given string.

    Parameters:
    string (string): The string for counting uppercase and lowercase letters.

    Returns:
    A tuple containing the count of uppercase and lowercase letters in the string.
    """
    uppercase = 0
    lowercase = 0

    for word in string:
        if word.isupper():
            uppercase += 1
        elif word.islower():
            lowercase += 1
    return uppercase, lowercase

print(count_case(string))
````

Example:

*Input: "Hello World"* 

*Expected Output: Uppercase count: 2, Lowercase count: 8* 

````python
import string
````

````python

sentence = "Note : this is an example !!! Good day : )"
def remove_punctuation(sentence):
    """
    Removes all punctuation marks (commas, periods, exclamation marks, question marks) from a sentence.

    Parameters:
    sentence (str): A string representing a sentence.

    Returns:
    str: The sentence without any punctuation marks.
    """
    sentence_limpia = sentence.translate(str.maketrans('','', string.punctuation))
    return sentence_limpia

print(remove_punctuation(sentence))

def word_count(sentence):
    """
  Cuenta el número de palabras en una oración determinada. Para hacer esto correctamente, primero elimina la puntuación de la oración.
    Nota: Una palabra se define como una secuencia de caracteres separados por espacios. Podemos suponer que no habrá espacios iniciales ni finales en la oración de entrada.
    
    Parámetros:
    oración (cadena): una cadena que representa una oración.

    Devoluciones:
    int: el número de palabras de la oración.
    """
    sentence_limpia = sentence.translate(str.maketrans('','', string.punctuation))
    sentence2 = len(sentence_limpia.split())
    return sentence2

print(word_count(sentence))
````

## Challenge 2: Build a Calculator

En este ejercicio, creará una calculadora utilizando funciones de Python. La calculadora podrá realizar operaciones aritméticas básicas como suma, resta, multiplicación y división.

Instrucciones
- Definir cuatro funciones para suma, resta, multiplicación y división.
- Cada función debe tomar dos argumentos, realizar la operación aritmética respectiva y devolver el resultado.
- Definir otra función llamada "calcular" que tome tres argumentos: dos operandos y un operador.
- La función "calcular" debe utilizar una declaración condicional para determinar qué función aritmética llamar en función del argumento del operador.
- La función "calcular" debería llamar a la función aritmética apropiada y devolver el resultado.
- Pruebe su función "calcular" llamándola con diferentes parámetros de entrada.


````python
def sumar(a,b):
    sum = a+b
    return sum

def restar(a,b):
    resta = a-b
    return resta

def multiplicar(a, b):
    multiplicacion = a*b
    return multiplicacion

def dividir(a, b):
    division = a/b
    return division
````

````python
def calcular(num1, num2, operador):
    if operador == sumar:
        return sumar(num1, num2)
    
    elif operador == restar:
            return restar(num1, num2)
    
    elif operador == multiplicar:
            return multiplicar(num1, num2)
    
    elif operador == dividir:
                return dividir(num1, num2)
    else:
        print("error: operador no valido")
````

````python
print(calcular(3,6, dividir))
print(calcular(10, 6.9, sumar))
print(calcular(8,15, multiplicar))
````

### Bonus: args and kwargs

Actualice el ejercicio anterior para que permita sumar, restar y multiplicar más de 2 números.

La calculadora podrá realizar operaciones aritméticas básicas como suma, resta, multiplicación y división de varios números.

*Hint: use args or kwargs. Recommended external resource: [Args and Kwargs in Python](https://www.geeksforgeeks.org/args-kwargs-python/)*

````python
def sumar(*args):
    result = 0
    for num in args:
        result += num
    return result

def restar(*args):
    result = 0
    for num in args:
        result -= num
    return result

def multiplicar(*args):
    result = 1
    for num in args:
        result *= num
    return result

def dividir(*args):
    result = args[0]
    for num in args[1:]:
        result /= num
    return result
````

````python
def calcular(operador, *args):
    if operador == sumar:
        return sumar(*args)
    
    elif operador == restar:
            return restar(*args)
    
    elif operador == multiplicar:
            return multiplicar(*args)
    
    elif operador == dividir:
                return dividir(*args)
    else:
        print("error: operador no valido")
````

````python
print(calcular(dividir, 3, 6, 9))
print(calcular(sumar, 10, 6.9, 10))
print(calcular(multiplicar, 8, 15, 2))
````

Hay varias formas de importar funciones desde un módulo de Python, como funciones.py, a un Jupyter Notebook:

1. Importar funciones específicas: si solo necesita usar algunas funciones del módulo, puede importarlas individualmente usando la palabra clave from. De esta manera, puedes llamar a las funciones directamente usando sus nombres, sin tener que usar el nombre del módulo. Por ejemplo:

```pitón
desde archivo_función importar nombre_función

nombre_función(argumentos)```

2. Importar el módulo completo: puede importar el módulo completo usando la palabra clave import seguida del nombre del módulo. Luego, puede llamar a las funciones usando la sintaxis module_name.function_name(). Ejemplo:

```pitón
importar archivo_función

archivo_función.nombre_función()
```

3. Cambiar el nombre de las funciones durante la importación: también puede cambiar el nombre de las funciones durante la importación utilizando la palabra clave as. Esto es útil si desea utilizar un nombre más corto o más descriptivo para la función en su código. Por ejemplo:

```pitón
desde archivo_función importar nombre_función como f

f.nombre_función(argumentos)
```

Independientemente del método que elija, asegúrese de que el archivo funciones.py esté en el mismo directorio que su Jupyter Notebook, o especifique la ruta al archivo en la declaración de importación o usando `sys.path`.
